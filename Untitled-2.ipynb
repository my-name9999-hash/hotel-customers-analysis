```python
import pandas as pd

# ----------Cleaning Data------------
# 1️⃣ Load Data
# -------------------------------
df = pd.read_csv(r"C:\Users\Esraa\Downloads\HotelCustomersDataset.csv")

# -------------------------------
# 2️⃣ Initial Exploration
# -------------------------------
print("Initial Shape:", df.shape)
print(df.info())
print(df.isnull().sum())
print(df.describe())

# -------------------------------
# 3️⃣ Remove Logically Invalid Data
# -------------------------------
if "Age" in df.columns:
    df = df[df["Age"] >= 0]

if "Children" in df.columns:
    df = df[df["Children"] >= 0]

if "BookingLeadTime" in df.columns:
    df = df[df["BookingLeadTime"] >= 0]

# -------------------------------
# 4️⃣ Handle Missing Values
# -------------------------------
# Drop rows with missing essential columns
essential_cols = ["Age", "AverageLeadTime", "BookingLeadTime"]
for col in essential_cols:
    if col in df.columns:
        df = df.dropna(subset=[col])

# Fill missing categorical values
categorical_cols = ["Nationality", "MarketSegment", "RepeatedGuest", "DistributionChannel"]
for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")

# Fill missing numeric values with median
numeric_cols = ["Children", "Babies", "ADR"]
for col in numeric_cols:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())

# -------------------------------
# 5️⃣ Remove Duplicates
# -------------------------------
df = df.drop_duplicates()
if "ID" in df.columns:
    df = df.drop_duplicates(subset="ID")

# -------------------------------
# 6️⃣ Fix Data Types
# -------------------------------
if "ID" in df.columns:
    df["ID"] = df["ID"].astype(str)

# Categorical columns
for col in ["MarketSegment", "RepeatedGuest", "Nationality", "DistributionChannel", "DepositType"]:
    if col in df.columns:
        df[col] = df[col].astype("category")

# Date columns
date_cols = ["BookingDate", "ArrivalDate", "DepartureDate"]
for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

# -------------------------------
# 7️⃣ Handle Outliers (IQR Method)
# -------------------------------
numeric_outlier_cols = ["AverageLeadTime", "ADR", "Children", "Babies"]
for col in numeric_outlier_cols:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df = df[(df[col] >= lower) & (df[col] <= upper)]

# -------------------------------
# 8️⃣ Final Checks
# -------------------------------
print("After Cleaning Shape:", df.shape)
print(df.isnull().sum())
print(df.dtypes)

# -------------------------------
# 9️⃣ Save Cleaned Dataset
# -------------------------------
df.to_csv("HotelCustomers_cleaned.csv", index=False, encoding="utf-8-sig")
print("✅ Data cleaned and saved successfully!")
```

Data Cleaning Phase (Customized for A Hotels Customers Dataset)
1.	Load Data
o	Imported the Excel file HotelCustomersDataset.xlsx into a pandas DataFrame.
2.	Initial Exploration
o	Checked the shape, data types, descriptive statistics, and missing values.
3.	Removing Invalid Records
o	Removed rows where Age, Children, or BookingLeadTime are negative, as these are not logically correct.
4.	Handling Missing Values
o	Dropped rows missing essential numeric columns: Age, AverageLeadTime, BookingLeadTime.
o	For categorical columns like Nationality, MarketSegment, RepeatedGuest, or DistributionChannel, filled missing values with "Unknown".
o	For other numeric columns like Children, Babies, and ADR, filled missing values with the median.
5.	Removing Duplicates
o	Dropped fully duplicate rows.
o	Optionally, removed duplicates based on unique ID.
6.	Fixing Data Types
o	Converted ID to string, since it's a unique identifier.
o	Converted categorical-like columns (MarketSegment, RepeatedGuest, Nationality, DistributionChannel, DepositType) to pandas category dtype.
o	Converted date columns (BookingDate, ArrivalDate, DepartureDate) to pandas datetime dtype.
7.	Handling Outliers
o	Used the IQR (Interquartile Range) method to remove outliers in numeric columns: AverageLeadTime, ADR, Children, and Babies to reduce skew and extreme values.
8.	Final Clean Check
o	Verified the final shape, missing values, and data types.
o	(Optional) Saved a cleaned version of the dataset as HotelCustomers_cleaned.csv for future use.
9.	Additional Notes
o	Ensured text consistency for categorical columns, e.g., Nationality is standardized using .str.upper().str.strip().
o	The dataset is now ready for analysis or machine learning pipelines, with numeric, categorical, and date columns properly prepared.
o	Using category dtype for categorical columns improves memory efficiency for large datasets.


```python
# =========================================
# Complete EDA - Hotel Customers Dataset
# =========================================

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Settings
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10,6)

# Load cleaned dataset
df = pd.read_csv('HotelCustomers_cleaned.csv')
print("Columns in dataset:", df.columns)
df.head()

# -----------------------------------------
# 1. Age distribution
plt.figure()
sns.histplot(df['Age'], bins=30, kde=True, color='skyblue')
plt.title('Distribution of Guests Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# -----------------------------------------
# 2. Top Nationalities
plt.figure()
top_nationalities = df['Nationality'].value_counts().nlargest(10)
sns.barplot(x=top_nationalities.values, y=top_nationalities.index, palette='viridis')
plt.title('Top 10 Nationalities')
plt.xlabel('Number of Guests')
plt.ylabel('Nationality')
plt.show()

# -----------------------------------------
# 3. Market Segment Pie
plt.figure()
market_counts = df['MarketSegment'].value_counts()
plt.pie(market_counts, labels=market_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Market Segment Distribution')
plt.show()

# -----------------------------------------
# 4. Average Lead Time Boxplot
plt.figure()
sns.boxplot(x=df['AverageLeadTime'], color='lightgreen')
plt.title('Boxplot of Average Lead Time')
plt.xlabel('Average Lead Time (days)')
plt.show()

# -----------------------------------------
# 5. Lodging Revenue
plt.figure()
sns.histplot(df['LodgingRevenue'], bins=30, kde=True, color='salmon')
plt.title('Distribution of Lodging Revenue')
plt.xlabel('Lodging Revenue')
plt.ylabel('Count')
plt.show()

# -----------------------------------------
# 6. Room Nights
plt.figure()
sns.histplot(df['RoomNights'], bins=30, kde=False, color='purple')
plt.title('Distribution of Room Nights')
plt.xlabel('Room Nights')
plt.ylabel('Count')
plt.show()

# -----------------------------------------
# 7. Special Requests Summary (SR columns)
sr_cols = [
    'SRHighFloor', 'SRLowFloor', 'SRAccessibleRoom', 'SRMediumFloor',
    'SRBathtub', 'SRShower', 'SRCrib', 'SRKingSizeBed', 'SRTwinBed',
    'SRNearElevator', 'SRAwayFromElevator', 'SRNoAlcoholInMiniBar', 'SRQuietRoom'
]
sr_counts = df[sr_cols].sum().sort_values(ascending=False)

plt.figure()
sns.barplot(x=sr_counts.values, y=sr_counts.index, palette='viridis')
plt.title('Special Requests Count')
plt.xlabel('Number of Bookings')
plt.ylabel('Special Request Type')
plt.show()

# -----------------------------------------
# 8. Bivariate Analysis Examples

# Age vs AverageLeadTime
plt.figure()
sns.scatterplot(x='Age', y='AverageLeadTime', data=df, alpha=0.5)
plt.title('Age vs Average Lead Time')
plt.xlabel('Age')
plt.ylabel('Average Lead Time')
plt.show()

# LodgingRevenue vs RoomNights
plt.figure()
sns.scatterplot(x='RoomNights', y='LodgingRevenue', data=df, alpha=0.5)
plt.title('Room Nights vs Lodging Revenue')
plt.xlabel('Room Nights')
plt.ylabel('Lodging Revenue')
plt.show()

# -----------------------------------------
# 9. Correlation Heatmap (Numeric Columns Only)
numeric_cols = df.select_dtypes(include='number')

plt.figure(figsize=(12,8))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap (Numeric Columns Only)')
plt.show()

# =========================================
# ✅ Notes:
# - This EDA includes univariate and bivariate analysis
# - Special requests are summarized in a bar plot
# - Correlation heatmap uses only numeric columns to avoid errors
# - All plots include titles, labels, and legends where appropriate
# =========================================
```
