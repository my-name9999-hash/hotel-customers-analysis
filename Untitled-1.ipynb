```python
# =====================
# ⭐ Step 0 — Load the dataset
# =====================
import pandas as pd
# =====================
# Load Dataset
# =====================
df = pd.read_csv(r"C:\Users\Esraa\Downloads\HotelCustomersDataset.csv")
# =====================
# Data Overview
# =====================
print("Shape of the dataset:", df.shape)
display(df.head())
# =====================
```

```python
#⭐ Step 1 — Initial Inspection
df.shape
df.info()
df.describe()
df.isnull().sum()
df.dtypes

```

 Markdown to paste:
Initial Dataset Assessment
Inspected the dataset shape (number of rows and columns).
Reviewed data types for each column.
Checked for missing values across all variables.
Generated statistical summaries to detect incorrect or extreme values.

```python
#Step 2 — Removing Invalid Values (Negative Ages)
#⛔ Issue:
#This dataset contains negative values in Age, which are impossible.
# Check negative ages
df[df["Age"] < 0]
# Remove rows with invalid negative ages
df = df[df["Age"] >= 0]
```

Fixing Invalid Values:
Detected negative values in the Age column.
Negative ages are logically impossible, so these rows were removed to maintain data quality.

```python
#⭐ Step 3 — Handling Missing Values
#First, check missing values:
df.isnull().sum()
#Example cleaning actions:
# Drop rows missing age or lead time
df = df.dropna(subset=["Age", "AverageLeadTime"])
```

Handling Missing Values:
Identified columns containing missing values.
Removed rows missing essential fields such as Age and AverageLeadTime.
Columns with high missing percentages and low importance were removed.
Ensured the dataset does not contain missing values that affect analysis.

```python
#⭐ Step 4 — Removing Duplicate
df = df.drop_duplicates()
```

Removing Duplicate Records:
Checked for duplicate rows.
Removed duplicates to avoid bias and ensure accuracy in analysis.

```python
#⭐ Step 5 — Fixing Data Types
#Some columns should be converted to the correct type.
df["ID"] = df["ID"].astype(str)
df["MarketSegment"] = df["MarketSegment"].astype("category")
#df["RepeatGuest"] = df["RepeatGuest"].astype("category")
```

Fixing Data Types:
Converted ID fields to string since they are identifiers, not numeric values.
Converted categorical variables into "category" type to optimize performance.
Ensured all numerical variables use proper numeric types (int/float).

```python
#Step 6 — Handling Outliers
#The most common outliers are in AverageLeadTime.
Q1 = df["AverageLeadTime"].quantile(0.25)
Q3 = df["AverageLeadTime"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df = df[(df["AverageLeadTime"] >= lower) & (df["AverageLeadTime"] <= upper)]
```

Handling Outliers:
Applied the Interquartile Range (IQR) method to detect and remove outliers in AverageLeadTime.
Outliers were removed because they distort the distribution and negatively affect analysis.

```python
# Final Clean Data Check
df.info()
df.describe()
df.isnull().sum()
df.shape
```

Final Cleaning Verification
Confirmed no remaining missing values.
Ensured all data types were corrected.
Verified the removal of invalid values, duplicates, and outliers.
The dataset is now fully cleaned and ready for EDA.
